# 1. 평가 지표 선택 이유 (Metric Selection)
이 테스트 데이터는 스팸 메일 분류기라고 가정했습니다. 샘플 데이터가 작고 직관적인 맞고 틀림을 확인하기 위해 **정확도(Accuracy)**를 평가 지표로 선택했습니다.

```python
# 2. 가짜 데이터 생성
y_true = [1, 0, 1, 1, 0, 0, 1, 0, 1, 1] 
y_pred = [1, 0, 0, 1, 0, 0, 1, 1, 1, 1] 

# 3. infer-ci를 사용한 신뢰구간 평가
import infer_ci
result = infer_ci.calculate_ci(y_true, y_pred, metric='accuracy')
print("신뢰구간 계산 결과:", result)
```

# 4. 결과 해석 (Interpretation)
계산 결과 나온 신뢰구간 하한선과 상한선은 이 모델이 새로운 데이터를 만났을 때 보여줄 수 있는 정확도의 최솟값과 최댓값을 의미합니다. 샘플이 매우 적기 때문에 구간이 다소 넓게 측정되었습니다.
