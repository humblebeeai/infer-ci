```python
# 1. infer-ci 라이브러리 및 필요 함수 가져오기
from infer_ci import f1_score

# 2. 아주 작은 크기의 샘플 데이터 준비 (정답과 모델 예측값)
y_true = [0, 1, 1, 0, 1, 0, 1, 1, 0, 1]
y_pred = [0, 1, 0, 0, 1, 0, 1, 0, 0, 1]

# 3. infer-ci를 활용한 F1-Score 평가 수행
f1, ci = f1_score(y_true, y_pred, confidence_interval=0.95, average='binary')

# 4. 결과 출력
print("=== Evaluation Results ===")
print(f"F1-Score: {f1}")
print(f"Confidence Interval: {ci}")
```
